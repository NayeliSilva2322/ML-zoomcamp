```python
def vector_multiplicacion(u,v):
    assert u.shape[0]==v.shape[0]
    n=u.shape[0]
    result=0.0
    
    for i in range(n): #n-1 es el tamaño real de la matriz 
        result=result+u[i]*v[i]
        
    return result
    
```

```python
import numpy as np
u=np.array([1,2,3])
v=np.array([9,8,7])
```

```python
vector_multiplicacion(u,v)
```

```python
#el numero de columnas de U debe de coincidir con el numero de filas v 
def matrix_vector_multiplicacion(U,v):
    assert U.shape[1]== v.shape[0]
    num_rows=U.shape[0]
    result=np.zeros(num_rows)
    for i in range(num_rows):
        result[i]=vector_multiplicacion(U[i], v)
    return result
```

```python
def matrix_matrix_multplicación(U,V):
    assert U.shape[1]== V.shape[0]
    num_rows=U.shape[1]
    num_colums=V.shape[0]
    print(num_rows,num_colums)
    result=np.zeros((num_rows, num_colums))
    for i in range(num_rows):
            vi = V[i, :]
            Uvi = vector_multiplicacion(U, vi) # Asegúrate de que esta función esté definida
            result[i, :] = Uvi
            
    return result
```
